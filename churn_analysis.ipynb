```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('data/WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
print(df.head())
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Contract', hue='Churn', data=df, palette='Set2')
plt.title('Churn Count by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Customer Count')

# This line automatically saves the chart into your folder!
plt.savefig('eda_charts.png', bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    x='Churn',
    y='MonthlyCharges',
    hue='Churn',
    data=df,
    palette='Pastel1',
    legend=False,
)
plt.title('Monthly Charges vs. Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Monthly Charges')

# Saves this chart with a distinct file name
plt.savefig('monthly_charges_churn.png', bbox_inches='tight')

plt.show()
```
